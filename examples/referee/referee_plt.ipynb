# 🤖 机器人任务评分规则依赖关系分析

本notebook分析机器人任务评分规则的依赖关系，并通过可视化图形展示任务之间的依赖结构。

## 📋 分析内容
1. **依赖关系图**: 展示任务间的前置条件和执行顺序
2. **统计分析**: 按类型、分值、依赖关系进行统计
3. **关键路径**: 识别核心任务和执行路径

## 🎯 任务类型
- 🔵 **2D位置检查**: 机器人位置验证
- 🟣 **3D位置检查**: 物体放置位置验证  
- 🟢 **抓取动作**: 物体抓取操作
- 🔴 **碰撞检查**: 安全碰撞检测

## 📊 评分规则数据
总共包含 **21个任务**，总分值 **160分**

```python
# 机器人任务评分规则依赖关系图分析
import zhplot
import matplotlib.pyplot as plt
import networkx as nx
import numpy as np
from collections import defaultdict

# 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# 定义节点颜色映射
type_colors = {
    'in_2d_range': '#3498db',      # 蓝色 - 2D位置检查
    'in_3d_range': '#9b59b6',      # 紫色 - 3D位置检查  
    'no_collision': '#e74c3c',     # 红色 - 碰撞检查
    'pick_up': '#2ecc71',          # 绿色 - 抓取动作
    'unknown': '#95a5a6'           # 灰色 - 未知类型
}

# 创建有向图
G = nx.DiGraph()

# 添加所有节点和依赖关系
node_labels = {}
node_colors = []
node_sizes = []
prerequisite_edges = []
until_edges = []

for rule in scoling_rules:
    description = rule['description']
    task_index = rule['task_index']
    score = rule['score']
    rule_type = rule.get('type', 'unknown')
    
    # 创建更详细的节点标签
    if '成功移动到桌子面前' in description:
        label = f"T{task_index}: 移动到桌子面前\n(2D位置检查, {score}分)"
    elif '从出发区移动到桌子面前物体抓取区过程中没有碰撞行为' in description:
        label = f"T{task_index}: 出发→桌子\n无碰撞({score}分)"
    elif '成功抓取普通物品 power_drill' in description:
        label = f"T{task_index}: 抓取电钻\n(抓取动作, {score}分)"
    elif '成功抓取普通物品 cube' in description:
        label = f"T{task_index}: 抓取方块\n(抓取动作, {score}分)"
    elif '成功抓取普通物品 extrusion' in description:
        label = f"T{task_index}: 抓取挤压件\n(抓取动作, {score}分)"
    elif '成功放置物品 power_drill' in description:
        label = f"T{task_index}: 放置电钻\n(3D位置, {score}分)"
    elif '成功放置物品 cube' in description:
        label = f"T{task_index}: 放置方块\n(3D位置, {score}分)"
    elif '成功放置物品 extrusion' in description:
        label = f"T{task_index}: 放置挤压件\n(3D位置, {score}分)"
    elif '抓取到放置中没有碰撞' in description:
        label = f"T{task_index}: 抓取放置过程\n无碰撞({score}分)"
    elif '成功抓取神秘物体' in description:
        label = f"T{task_index}: 抓取神秘物体\n★高价值★({score}分)"
    elif '成功把神秘物体放置到料框里' in description:
        label = f"T{task_index}: 神秘物体→料框\n(3D位置, {score}分)"
    elif '神秘物体放置到料框的过程中没有碰撞行为' in description:
        label = f"T{task_index}: 神秘物体放置\n无碰撞({score}分)"
    elif '成功移动到料框抓取区（桌子侧面）' in description:
        label = f"T{task_index}: 移动到料框区\n(桌子侧面, {score}分)"
    elif '从物体抓取区移动到料框抓取区过程中没有碰撞行为' in description:
        label = f"T{task_index}: 抓取区→料框区\n无碰撞({score}分)"
    elif '成功抓起料框' in description:
        label = f"T{task_index}: 抓取料框\n(抓取动作, {score}分)"
    elif '成功移动到货架面前' in description:
        label = f"T{task_index}: 移动到货架前\n(2D位置, {score}分)"
    elif '从料框抓取区移动到料框放置区（货架）过程中没有碰撞行为' in description:
        label = f"T{task_index}: 料框区→货架\n无碰撞({score}分)"
    elif '成功将箱子放置到货架上' in description:
        label = f"T{task_index}: 料框放置货架\n(3D位置, {score}分)"
    elif '将箱子放置到货架的过程中没有碰撞行为' in description:
        label = f"T{task_index}: 货架放置过程\n无碰撞({score}分)"
    elif '成功移动到任务结束点' in description:
        label = f"T{task_index}: 移动到结束点\n(2D位置, {score}分)"
    elif '从料框放置区移动到任务结束区过程中没有碰撞行为' in description:
        label = f"T{task_index}: 货架→结束点\n无碰撞({score}分)"
    else:
        label = f"T{task_index}: {description[:10]}...\n({score}分)"
    
    # 添加节点
    G.add_node(description)
    node_labels[description] = label
    node_colors.append(type_colors.get(rule_type, '#95a5a6'))
    # 根据分数调整节点大小，神秘物体任务特别突出
    if score == 30:
        node_sizes.append(800)  # 神秘物体任务特别大
    else:
        node_sizes.append(500 + score * 20)
    
    # 添加前置依赖边
    if 'prerequisite' in rule:
        for prereq in rule['prerequisite']:
            if prereq in [r['description'] for r in scoling_rules]:
                G.add_edge(prereq, description)
                prerequisite_edges.append((prereq, description))
    
    # 添加until边
    if 'until' in rule:
        for until_condition in rule['until']:
            if until_condition in [r['description'] for r in scoling_rules]:
                until_edges.append((description, until_condition))

print(f"✅ 成功创建图：{G.number_of_nodes()} 个节点，{G.number_of_edges()} 条前置依赖边，{len(until_edges)} 条until边")

# 创建单个层次图
fig, ax = plt.subplots(figsize=(6, 24))

# 层次布局 - 按任务阶段分组
task_groups = defaultdict(list)
for rule in scoling_rules:
    task_groups[rule['task_index']].append(rule)

pos = {}
# 重新设计Y轴位置，更好地反映任务流程
y_positions = {
    1: 10,   # 开始：移动到桌子前
    2: 9,    # 移动碰撞检查
    3: 8,    # 抓取普通物品
    4: 7,    # 放置普通物品
    5: 6,    # 抓取放置碰撞检查
    6: 5,    # 抓取神秘物体
    7: 4,    # 放置神秘物体
    8: 3,    # 神秘物体碰撞检查
    9: 2,    # 移动到料框区
    10: 1,   # 移动碰撞检查
    11: 0,   # 抓取料框
    13: -1,  # 移动到货架
    14: -2,  # 料框移动碰撞检查
    15: -3,  # 放置料框到货架
    16: -4,  # 货架放置碰撞检查
    17: -5,  # 移动到结束点
    18: -6   # 最终移动碰撞检查
}

# 为每个任务组分配位置
for task_index, rules in task_groups.items():
    y = y_positions.get(task_index, 0)
    
    # 如果同一task_index有多个规则，水平排列
    if len(rules) == 1:
        pos[rules[0]['description']] = (0, y)
    else:
        x_spacing = 4  # 增加水平间距
        x_start = -(len(rules) - 1) * x_spacing / 2
        for i, rule in enumerate(rules):
            pos[rule['description']] = (x_start + i * x_spacing, y)

# 绘制节点
nx.draw_networkx_nodes(G, pos, node_color=node_colors, node_size=node_sizes,
                      alpha=0.8, linewidths=2, edgecolors='black', ax=ax)

# 绘制前置依赖边
nx.draw_networkx_edges(G, pos, edgelist=prerequisite_edges, edge_color='#2c3e50', 
                      arrows=True, arrowsize=25, arrowstyle='->', width=2, alpha=0.8,
                      connectionstyle="arc3,rad=0.15", ax=ax)

# 绘制until边（虚线）
if until_edges:
    nx.draw_networkx_edges(G, pos, edgelist=until_edges, edge_color='#e67e22', 
                          arrows=True, arrowsize=20, arrowstyle='->', width=2, alpha=0.7,
                          style='dashed', connectionstyle="arc3,rad=-0.15", ax=ax)

# 绘制详细标签
nx.draw_networkx_labels(G, pos, node_labels, font_size=9, font_weight='bold', ax=ax)

# 添加更详细的任务阶段标注
stage_labels = {
    10: "任务开始", 
    9: "移动阶段", 
    8: "抓取普通物品", 
    7: "放置普通物品", 
    6: "抓取过程检查", 
    5: "神秘物体抓取", 
    4: "神秘物体放置", 
    3: "神秘物体检查", 
    2: "移动到料框区", 
    1: "移动过程检查", 
    0: "料框抓取", 
    -1: "移动到货架", 
    -2: "料框移动检查", 
    -3: "货架放置", 
    -4: "货架放置检查", 
    -5: "移动到终点", 
    -6: "最终检查"
}

for y, label in stage_labels.items():
    ax.text(-12, y, label, fontsize=16, fontweight='bold', 
           bbox=dict(boxstyle="round,pad=0.5", facecolor='lightblue', alpha=0.8),
           verticalalignment='center')

# 添加分数统计文本框
score_info = f"""
分数统计:
• 总分: {sum(r['score'] for r in scoling_rules)}分
• 神秘物体: 30分 (最高)
• 抓取动作: {sum(r['score'] for r in scoling_rules if r.get('type') == 'pick_up')}分
• 位置检查: {sum(r['score'] for r in scoling_rules if 'range' in r.get('type', ''))}分
• 碰撞检查: {sum(r['score'] for r in scoling_rules if r.get('type') == 'no_collision')}分
"""

ax.text(12, 8, score_info, fontsize=16, 
        bbox=dict(boxstyle="round,pad=0.8", facecolor='lightyellow', alpha=0.9),
        verticalalignment='top')

# 创建详细图例
legend_elements = [
    plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='#3498db', markersize=15, label='2D位置检查'),
    plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='#9b59b6', markersize=15, label='3D位置检查'),
    plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='#2ecc71', markersize=15, label='抓取动作'),
    plt.Line2D([0], [0], marker='o', color='w', markerfacecolor='#e74c3c', markersize=15, label='碰撞检查'),
    plt.Line2D([0], [0], color='#2c3e50', linewidth=3, label='→ 前置依赖'),
    plt.Line2D([0], [0], color='#e67e22', linewidth=3, linestyle='--', label='--> Until条件')
]

ax.legend(handles=legend_elements, loc='upper right', bbox_to_anchor=(1, 1), fontsize=16,
         title="任务类型与关系", title_fontsize=18)

ax.set_title("机器人任务评分规则依赖关系图（层次化流程）", fontsize=24, fontweight='bold', pad=30)
ax.axis('off')

# 调整布局
plt.tight_layout()
plt.show()
```
